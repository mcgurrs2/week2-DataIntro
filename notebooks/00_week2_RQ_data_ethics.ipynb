## 1. Research Question

- **Main research question:** Does personality improve the prediction of stress in university students in addition to lifestyle and demographic factors?
- **Why is this question interesting or important?** Stress levels are often hightened in university due to many different factors including financial and academic stressors, however, it has been found that some individuals are more likely to experience stress than others despite having similar lifestyles and backgrounds (Luo et al., 2023). Stress has also been found to be linked with serious mental health conditions such as anxiety and depression (Liu, 2013).
- **Who might use or care about the answer?** The outcome could be very useful to mental health professionals on and off campus, such as college counsellors and student therapists, so that more useful interventions can be put into place to help students cope with stressors.
- **Possible sub-questions:** Do lifestyle and demographic factors predict stress levels more accurately than when personality traits are included? Which personality traits are associated with high stress levels?
